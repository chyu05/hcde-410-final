# Motivation/problem statement: Why are you planning to do this analysis?
My plan is to look at changes in ridership data for the Link Light Rail and to examine relevant events (station openings, maintenance, major events such as FIFA world cup) to see how ridership changes. I would say that this is interesting because instead of just looking at ridership trends, we can look at the reasoning behind the trends that you may see when looking at ridership data. Looking at the reasoning behind the ridership changes can also be beneficial for Sound Transit as it can allow them to see what changes help boost ridership and that can help inform their future decisions. As an avid transit rider, I mainly hope to learn about what causes increases and decreases in ridership.

# Data used: What dataset do you plan to use, and why?
I will use [Sound Transit's ridership data](https://www.soundtransit.org/ride-with-us/system-performance-tracker/ridership), with a ["personal, royalty-free, non-assignable, and non-exclusive license to use the Website Content in the United States only as an informative resource"](https://www.soundtransit.org/help-contacts/business-information/terms-use). The dataset shows monthly total and average boardings for each of Sound Transit's services. This dataset is Sound Transit's own ridership data published on their own site, making it the most relevant dataset to use for my project (not sure if there even are others). I wouldn't expect there to be any ethical issues with using this dataset as its use in the project is not commercial or malicious in any way.
